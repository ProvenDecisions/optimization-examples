## The problem

Demand is 20, 30 and 25 units in periods 1-3. Each production run costs \$100 to set up plus \$5 per unit; holding a unit for one period costs \$2. No starting or ending stock. When should you produce, and how much?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 s_{1} + 2 s_{2} + 5 x_{1} + 5 x_{2} + 5 x_{3} + 100 y_{1} + 100 y_{2} + 100 y_{3} &&\text{make this as small as possible}\\ \text{s.t.}\;& - s_{1} + x_{1} = 20 &&\text{limit 1: left side must equal 20}\\ \;& s_{1} - s_{2} + x_{2} = 30 &&\text{limit 2: left side must equal 30}\\ \;& s_{2} + x_{3} = 25 &&\text{limit 3: left side must equal 25}\\ \;& x_{1} \le 75 y_{1} &&\text{limit 4: left side cannot exceed the right side}\\ \;& x_{2} \le 55 y_{2} &&\text{limit 5: left side cannot exceed the right side}\\ \;& x_{3} \le 25 y_{3} &&\text{limit 6: left side cannot exceed the right side}\\ \;& x_{1} \ge 0 &&\text{units produced in period 1 (units) cannot be negative}\\ \;& x_{2} \ge 0 &&\text{units produced in period 2 (units) cannot be negative}\\ \;& x_{3} \ge 0 &&\text{units produced in period 3 (units) cannot be negative}\\ \;& s_{1} \ge 0 &&\text{inventory held at end of period 1 (units) cannot be negative}\\ \;& s_{2} \ge 0 &&\text{inventory held at end of period 2 (units) cannot be negative}\\ &y_{1}, y_{2}, y_{3} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1')  # units produced in period 1 (units)
s1 = cp.Variable(name='s1')  # inventory held at end of period 1 (units)
x2 = cp.Variable(name='x2')  # units produced in period 2 (units)
s2 = cp.Variable(name='s2')  # inventory held at end of period 2 (units)
x3 = cp.Variable(name='x3')  # units produced in period 3 (units)
y1 = cp.Variable(name='y1', boolean=True)  # 1 if a production run is set up in period 1, else 0
y2 = cp.Variable(name='y2', boolean=True)  # 1 if a production run is set up in period 2, else 0
y3 = cp.Variable(name='y3', boolean=True)  # 1 if a production run is set up in period 3, else 0

objective = cp.Minimize(2*s1 + 2*s2 + 5*x1 + 5*x2 + 5*x3 + 100*y1 + 100*y2 + 100*y3)
constraints = [
    -s1 + x1 == 20,
    s1 - s2 + x2 == 30,
    s2 + x3 == 25,
    x1 <= 75*y1,
    x2 <= 55*y2,
    x3 <= 25*y3,
    x1 >= 0,
    x2 >= 0,
    x3 >= 0,
    s1 >= 0,
    s2 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('s1 =', s1.value)
print('x2 =', x2.value)
print('s2 =', s2.value)
print('x3 =', x3.value)
print('y1 =', y1.value)
print('y2 =', y2.value)
print('y3 =', y3.value)
```

## The checked answer

You asked for: the production schedule (which periods to produce in and how much in each), along with the minimum total cost. The smallest objective value is **625**, at x1 = 20, s1 = 0, x2 = 55, s2 = 25, x3 = 0, y1 = 1, y2 = 1, y3 = 0. All 11 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**

**Problem class:** MILP; **solver:** HIGHS.
