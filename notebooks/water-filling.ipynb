## The problem

Split 1 watt of transmit power over four channels with noise levels of 0.2, 0.5, 0.9 and 1.5 watts to maximize the total capacity, the sum of log(1 + power_i / noise_i), with each power in watts. How much power goes to each channel?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& \log{\left(5.0 p_{1} + 1 \right)} + \log{\left(2.0 p_{2} + 1 \right)} + \log{\left(1.111111111 p_{3} + 1 \right)} + \log{\left(0.6666666667 p_{4} + 1 \right)} &&\text{make as large as possible: total capacity, sum of log(1 + power\_i / noise\_i) (nats)}\\ \text{s.t.}\;& p_{1} + p_{2} + p_{3} + p_{4} \le 1 &&\text{limit 1: left side cannot exceed 1}\\ \;& p_{1} \ge 0 &&\text{power allocated to channel 1 with noise 0.2 (watts) cannot be negative}\\ \;& p_{2} \ge 0 &&\text{power allocated to channel 2 with noise 0.5 (watts) cannot be negative}\\ \;& p_{3} \ge 0 &&\text{power allocated to channel 3 with noise 0.9 (watts) cannot be negative}\\ \;& p_{4} \ge 0 &&\text{power allocated to channel 4 with noise 1.5 (watts) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

p1 = cp.Variable(name='p1')  # power allocated to channel 1 with noise 0.2 (watts)
p2 = cp.Variable(name='p2')  # power allocated to channel 2 with noise 0.5 (watts)
p3 = cp.Variable(name='p3')  # power allocated to channel 3 with noise 0.9 (watts)
p4 = cp.Variable(name='p4')  # power allocated to channel 4 with noise 1.5 (watts)

objective = cp.Maximize(cp.log(5.0*p1 + 1) + cp.log(2.0*p2 + 1) + cp.log(1.11111111111111*p3 + 1) + cp.log(0.666666666666667*p4 + 1))
constraints = [
    p1 + p2 + p3 + p4 <= 1,
    p1 >= 0,
    p2 >= 0,
    p3 >= 0,
    p4 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('p1 =', p1.value)
print('p2 =', p2.value)
print('p3 =', p3.value)
print('p4 =', p4.value)
```

## The checked answer

You asked for: the power allocated to each channel (p1, p2, p3, p4). The largest objective value is **1.97755**, at p1 = 0.649973, p2 = 0.350027, p3 = 5.77313e-08, p4 = 3.28869e-09. All 5 constraints are satisfied (violation ≤0e+00). **Optimal, numerically checked.**

**Problem class:** Conic (convex); **solver:** CLARABEL.
