## The problem

A buyer orders x tonnes of grade A and y tonnes of grade B (x, y >= 0) at \$3 and \$2 per tonne, and must take at least 1 tonne of grade B. The contract is met if either the total order is at least 6 tonnes, or at least 3 tonnes of grade A are ordered (or both). Minimize cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 3 x + 2 y &&\text{make as small as possible: total purchase cost (dollars)}\\ \text{s.t.}\;& y \ge 1 &&\text{tonnes of grade B ordered (tonnes) is at least 1}\\ \;& x + y \ge 6 w &&\text{limit 2: left side must reach at least the right side}\\ \;& x \ge 3 - 3 w &&\text{limit 3: left side must reach at least the right side}\\ \;& x \ge 0 &&\text{tonnes of grade A ordered (tonnes) cannot be negative}\\ \;& y \ge 0 &&\text{tonnes of grade B ordered (tonnes) cannot be negative}\\ &w \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

y = cp.Variable(name='y')  # tonnes of grade B ordered (tonnes)
x = cp.Variable(name='x')  # tonnes of grade A ordered (tonnes)
w = cp.Variable(name='w', boolean=True)  # 1 if the contract is met via total order >= 6 tonnes, 0 if via grade A >= 3 tonnes (yes/no)

objective = cp.Minimize(3*x + 2*y)
constraints = [
    y >= 1,
    x + y >= 6*w,
    x >= 3 - 3*w,
    x >= 0,
    y >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('y =', y.value)
print('x =', x.value)
print('w =', w.value)
```

## The checked answer

Assumed: "either ... or (or both)" is modeled with a binary switch w. When w=1, x+y>=6 is enforced. When w=0, x>=3 is enforced. Both can still hold when one is selected. The smallest objective value is **11**, at y = 1, x = 3, w = 0. All 5 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
