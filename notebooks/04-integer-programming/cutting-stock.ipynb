## The problem

Rolls are 10 m wide. Customers need four pieces 3 m wide and three pieces 4 m wide. What is the fewest rolls to cut?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& n_{1} + n_{2} + n_{3} &&\text{make as small as possible: total number of 10 m rolls cut (rolls)}\\ \text{s.t.}\;& 3 n_{1} + 2 n_{2} \ge 4 &&\text{limit 1: left side must reach at least 4}\\ \;& n_{2} + 2 n_{3} \ge 3 &&\text{limit 2: left side must reach at least 3}\\ \;& n_{1} \ge 0 &&\text{rolls cut with pattern A: three 3 m pieces (9 m used) (rolls) cannot be negative}\\ \;& n_{2} \ge 0 &&\text{rolls cut with pattern B: two 3 m pieces and one 4 m piece (10 m used) (rolls) cannot be negative}\\ \;& n_{3} \ge 0 &&\text{rolls cut with pattern C: two 4 m pieces (8 m used) (rolls) cannot be negative}\\ &n_{1}, n_{2}, n_{3} \in \mathbb{Z} &&\text{whole numbers only}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

n1 = cp.Variable(name='n1', integer=True)  # rolls cut with pattern A: three 3 m pieces (9 m used) (rolls)
n2 = cp.Variable(name='n2', integer=True)  # rolls cut with pattern B: two 3 m pieces and one 4 m piece (10 m used) (rolls)
n3 = cp.Variable(name='n3', integer=True)  # rolls cut with pattern C: two 4 m pieces (8 m used) (rolls)

objective = cp.Minimize(n1 + n2 + n3)
constraints = [
    3*n1 + 2*n2 >= 4,
    n2 + 2*n3 >= 3,
    n1 >= 0,
    n2 >= 0,
    n3 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('n1 =', n1.value)
print('n2 =', n2.value)
print('n3 =', n3.value)
```

## The checked answer

Assumed: only these maximal cutting patterns are needed, since any other pattern is dominated by one of them; extra pieces beyond demand are allowed (>=). The smallest objective value is **3**, at n1 = 0, n2 = 2, n3 = 1. All 5 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
