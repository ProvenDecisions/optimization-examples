## The problem

maximize x + y
subject to
  2x + 2y <= 3
  x, y >= 0
integer: x, y

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& x + y &&\text{make as large as possible: sum of x and y (units)}\\ \text{s.t.}\;& 2 x + 2 y \le 3 &&\text{limit 1: left side cannot exceed 3}\\ \;& x \ge 0 &&\text{first decision variable (units) cannot be negative}\\ \;& y \ge 0 &&\text{second decision variable (units) cannot be negative}\\ &x, y \in \mathbb{Z} &&\text{whole numbers only}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x = cp.Variable(name='x', integer=True)  # first decision variable (units)
y = cp.Variable(name='y', integer=True)  # second decision variable (units)

objective = cp.Maximize(x + y)
constraints = [
    2*x + 2*y <= 3,
    x >= 0,
    y >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x =', x.value)
print('y =', y.value)
```

## The checked answer

The largest objective value is **1**, at x = 1, y = 0. All 3 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
