## The problem

Colour the vertices of a five-cycle (edges 1-2, 2-3, 3-4, 4-5, 5-1) so adjacent vertices differ, using as few colours as possible.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& y_{1} + y_{2} + y_{3} &&\text{make this as small as possible}\\ \text{s.t.}\;& x_{1 1} + x_{1 2} + x_{1 3} = 1 &&\text{limit 1: left side must equal 1}\\ \;& x_{2 1} + x_{2 2} + x_{2 3} = 1 &&\text{limit 2: left side must equal 1}\\ \;& x_{3 1} + x_{3 2} + x_{3 3} = 1 &&\text{limit 3: left side must equal 1}\\ \;& x_{4 1} + x_{4 2} + x_{4 3} = 1 &&\text{limit 4: left side must equal 1}\\ \;& x_{5 1} + x_{5 2} + x_{5 3} = 1 &&\text{limit 5: left side must equal 1}\\ \;& x_{1 1} + x_{2 1} \le y_{1} &&\text{limit 6: left side cannot exceed the right side}\\ \;& x_{1 2} + x_{2 2} \le y_{2} &&\text{limit 7: left side cannot exceed the right side}\\ \;& x_{1 3} + x_{2 3} \le y_{3} &&\text{limit 8: left side cannot exceed the right side}\\ \;& x_{2 1} + x_{3 1} \le y_{1} &&\text{limit 9: left side cannot exceed the right side}\\ \;& x_{2 2} + x_{3 2} \le y_{2} &&\text{limit 10: left side cannot exceed the right side}\\ \;& x_{2 3} + x_{3 3} \le y_{3} &&\text{limit 11: left side cannot exceed the right side}\\ \;& x_{3 1} + x_{4 1} \le y_{1} &&\text{limit 12: left side cannot exceed the right side}\\ \;& x_{3 2} + x_{4 2} \le y_{2} &&\text{limit 13: left side cannot exceed the right side}\\ \;& x_{3 3} + x_{4 3} \le y_{3} &&\text{limit 14: left side cannot exceed the right side}\\ \;& x_{4 1} + x_{5 1} \le y_{1} &&\text{limit 15: left side cannot exceed the right side}\\ \;& x_{4 2} + x_{5 2} \le y_{2} &&\text{limit 16: left side cannot exceed the right side}\\ \;& x_{4 3} + x_{5 3} \le y_{3} &&\text{limit 17: left side cannot exceed the right side}\\ \;& x_{1 1} + x_{5 1} \le y_{1} &&\text{limit 18: left side cannot exceed the right side}\\ \;& x_{1 2} + x_{5 2} \le y_{2} &&\text{limit 19: left side cannot exceed the right side}\\ \;& x_{1 3} + x_{5 3} \le y_{3} &&\text{limit 20: left side cannot exceed the right side}\\ &x_{1 1}, x_{1 2}, x_{1 3}, x_{2 1}, x_{2 2}, x_{2 3}, x_{3 1}, x_{3 2}, x_{3 3}, x_{4 1}, x_{4 2}, x_{4 3}, x_{5 1}, x_{5 2}, x_{5 3}, y_{1}, y_{2}, y_{3} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1_1 = cp.Variable(name='x1_1', boolean=True)  # 1 if vertex 1 gets colour 1 (yes/no); the other xi_c variables follow the same pattern for vertex i and colour c
x1_2 = cp.Variable(name='x1_2', boolean=True)  # see x1_1
x1_3 = cp.Variable(name='x1_3', boolean=True)  # see x1_1
x2_1 = cp.Variable(name='x2_1', boolean=True)  # see x1_1
x2_2 = cp.Variable(name='x2_2', boolean=True)  # see x1_1
x2_3 = cp.Variable(name='x2_3', boolean=True)  # see x1_1
x3_1 = cp.Variable(name='x3_1', boolean=True)  # see x1_1
x3_2 = cp.Variable(name='x3_2', boolean=True)  # see x1_1
x3_3 = cp.Variable(name='x3_3', boolean=True)  # see x1_1
x4_1 = cp.Variable(name='x4_1', boolean=True)  # see x1_1
x4_2 = cp.Variable(name='x4_2', boolean=True)  # see x1_1
x4_3 = cp.Variable(name='x4_3', boolean=True)  # see x1_1
x5_1 = cp.Variable(name='x5_1', boolean=True)  # see x1_1
x5_2 = cp.Variable(name='x5_2', boolean=True)  # see x1_1
x5_3 = cp.Variable(name='x5_3', boolean=True)  # see x1_1
y1 = cp.Variable(name='y1', boolean=True)  # 1 if colour 1 is used (yes/no)
y2 = cp.Variable(name='y2', boolean=True)  # 1 if colour 2 is used (yes/no)
y3 = cp.Variable(name='y3', boolean=True)  # 1 if colour 3 is used (yes/no)

objective = cp.Minimize(y1 + y2 + y3)
constraints = [
    x1_1 + x1_2 + x1_3 == 1,
    x2_1 + x2_2 + x2_3 == 1,
    x3_1 + x3_2 + x3_3 == 1,
    x4_1 + x4_2 + x4_3 == 1,
    x5_1 + x5_2 + x5_3 == 1,
    x1_1 + x2_1 <= y1,
    x1_2 + x2_2 <= y2,
    x1_3 + x2_3 <= y3,
    x2_1 + x3_1 <= y1,
    x2_2 + x3_2 <= y2,
    x2_3 + x3_3 <= y3,
    x3_1 + x4_1 <= y1,
    x3_2 + x4_2 <= y2,
    x3_3 + x4_3 <= y3,
    x4_1 + x5_1 <= y1,
    x4_2 + x5_2 <= y2,
    x4_3 + x5_3 <= y3,
    x1_1 + x5_1 <= y1,
    x1_2 + x5_2 <= y2,
    x1_3 + x5_3 <= y3,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1_1 =', x1_1.value)
print('x1_2 =', x1_2.value)
print('x1_3 =', x1_3.value)
print('x2_1 =', x2_1.value)
print('x2_2 =', x2_2.value)
print('x2_3 =', x2_3.value)
print('x3_1 =', x3_1.value)
print('x3_2 =', x3_2.value)
print('x3_3 =', x3_3.value)
print('x4_1 =', x4_1.value)
print('x4_2 =', x4_2.value)
print('x4_3 =', x4_3.value)
print('x5_1 =', x5_1.value)
print('x5_2 =', x5_2.value)
print('x5_3 =', x5_3.value)
print('y1 =', y1.value)
print('y2 =', y2.value)
print('y3 =', y3.value)
```

All 20 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
