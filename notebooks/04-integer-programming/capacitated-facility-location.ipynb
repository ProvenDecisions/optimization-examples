## The problem

Two depots cost \$20 and \$25 to open and can ship 30 and 25 units. Three stores need 15, 20 and 10 units. Shipping one unit costs \$2, \$4 and \$5 from depot 1 to stores 1, 2 and 3, and \$5, \$3 and \$2 from depot 2. Minimize opening plus shipping cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 x_{11} + 4 x_{12} + 5 x_{13} + 5 x_{21} + 3 x_{22} + 2 x_{23} + 20 y_{1} + 25 y_{2} &&\text{make as small as possible: total depot opening cost plus shipping cost (\$)}\\ \text{s.t.}\;& x_{11} + x_{12} + x_{13} \le 30 y_{1} &&\text{limit 1: left side cannot exceed the right side}\\ \;& x_{21} + x_{22} + x_{23} \le 25 y_{2} &&\text{limit 2: left side cannot exceed the right side}\\ \;& x_{11} + x_{21} = 15 &&\text{limit 3: left side must equal 15}\\ \;& x_{12} + x_{22} = 20 &&\text{limit 4: left side must equal 20}\\ \;& x_{13} + x_{23} = 10 &&\text{limit 5: left side must equal 10}\\ \;& x_{11} \ge 0 &&\text{units shipped from depot 1 to store 1 (units) cannot be negative}\\ \;& x_{12} \ge 0 &&\text{units shipped from depot 1 to store 2 (units) cannot be negative}\\ \;& x_{13} \ge 0 &&\text{units shipped from depot 1 to store 3 (units) cannot be negative}\\ \;& x_{21} \ge 0 &&\text{units shipped from depot 2 to store 1 (units) cannot be negative}\\ \;& x_{22} \ge 0 &&\text{units shipped from depot 2 to store 2 (units) cannot be negative}\\ \;& x_{23} \ge 0 &&\text{units shipped from depot 2 to store 3 (units) cannot be negative}\\ &y_{1}, y_{2} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x11 = cp.Variable(name='x11')  # units shipped from depot 1 to store 1 (units)
x12 = cp.Variable(name='x12')  # units shipped from depot 1 to store 2 (units)
x13 = cp.Variable(name='x13')  # units shipped from depot 1 to store 3 (units)
y1 = cp.Variable(name='y1', boolean=True)  # 1 if depot 1 is opened, else 0 (yes/no)
x21 = cp.Variable(name='x21')  # units shipped from depot 2 to store 1 (units)
x22 = cp.Variable(name='x22')  # units shipped from depot 2 to store 2 (units)
x23 = cp.Variable(name='x23')  # units shipped from depot 2 to store 3 (units)
y2 = cp.Variable(name='y2', boolean=True)  # 1 if depot 2 is opened, else 0 (yes/no)

objective = cp.Minimize(2*x11 + 4*x12 + 5*x13 + 5*x21 + 3*x22 + 2*x23 + 20*y1 + 25*y2)
constraints = [
    x11 + x12 + x13 <= 30*y1,
    x21 + x22 + x23 <= 25*y2,
    x11 + x21 == 15,
    x12 + x22 == 20,
    x13 + x23 == 10,
    x11 >= 0,
    x12 >= 0,
    x13 >= 0,
    x21 >= 0,
    x22 >= 0,
    x23 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x11 =', x11.value)
print('x12 =', x12.value)
print('x13 =', x13.value)
print('y1 =', y1.value)
print('x21 =', x21.value)
print('x22 =', x22.value)
print('x23 =', x23.value)
print('y2 =', y2.value)
```

## The checked answer

The smallest objective value is **160**, at x11 = 15, x12 = 5, x13 = 0, y1 = 1, x21 = 0, x22 = 15, x23 = 10, y2 = 1. All 11 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
