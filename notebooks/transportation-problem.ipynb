## The problem

Plants 1 and 2 can ship 20 and 30 units. Markets A, B and C need 10, 25 and 15 units. Shipping one unit costs, in dollars: plant 1 to A, B, C = \$8, \$6, \$10; plant 2 to A, B, C = \$9, \$12, \$13. Find the shipping plan with the lowest total cost in dollars.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 8 x_{1A} + 6 x_{1B} + 10 x_{1C} + 9 x_{2A} + 12 x_{2B} + 13 x_{2C} &&\text{make as small as possible: total shipping cost (dollars)}\\ \text{s.t.}\;& x_{1A} + x_{1B} + x_{1C} \le 20 &&\text{limit 1: left side cannot exceed 20}\\ \;& x_{2A} + x_{2B} + x_{2C} \le 30 &&\text{limit 2: left side cannot exceed 30}\\ \;& x_{1A} + x_{2A} \ge 10 &&\text{limit 3: left side must reach at least 10}\\ \;& x_{1B} + x_{2B} \ge 25 &&\text{limit 4: left side must reach at least 25}\\ \;& x_{1C} + x_{2C} \ge 15 &&\text{limit 5: left side must reach at least 15}\\ \;& x_{1A} \ge 0 &&\text{units shipped from plant 1 to market A (units) cannot be negative}\\ \;& x_{1B} \ge 0 &&\text{units shipped from plant 1 to market B (units) cannot be negative}\\ \;& x_{1C} \ge 0 &&\text{units shipped from plant 1 to market C (units) cannot be negative}\\ \;& x_{2A} \ge 0 &&\text{units shipped from plant 2 to market A (units) cannot be negative}\\ \;& x_{2B} \ge 0 &&\text{units shipped from plant 2 to market B (units) cannot be negative}\\ \;& x_{2C} \ge 0 &&\text{units shipped from plant 2 to market C (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1A = cp.Variable(name='x1A')  # units shipped from plant 1 to market A (units)
x1B = cp.Variable(name='x1B')  # units shipped from plant 1 to market B (units)
x1C = cp.Variable(name='x1C')  # units shipped from plant 1 to market C (units)
x2A = cp.Variable(name='x2A')  # units shipped from plant 2 to market A (units)
x2B = cp.Variable(name='x2B')  # units shipped from plant 2 to market B (units)
x2C = cp.Variable(name='x2C')  # units shipped from plant 2 to market C (units)

objective = cp.Minimize(8*x1A + 6*x1B + 10*x1C + 9*x2A + 12*x2B + 13*x2C)
constraints = [
    x1A + x1B + x1C <= 20,
    x2A + x2B + x2C <= 30,
    x1A + x2A >= 10,
    x1B + x2B >= 25,
    x1C + x2C >= 15,
    x1A >= 0,
    x1B >= 0,
    x1C >= 0,
    x2A >= 0,
    x2B >= 0,
    x2C >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1A =', x1A.value)
print('x1B =', x1B.value)
print('x1C =', x1C.value)
print('x2A =', x2A.value)
print('x2B =', x2B.value)
print('x2C =', x2C.value)
```

## The checked answer

The smallest objective value is **465**, at x1A = 0, x1B = 20, x1C = 0, x2A = 10, x2B = 5, x2C = 15. All 11 constraints are satisfied exactly. **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
