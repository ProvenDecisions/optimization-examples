## The problem

Three warehouses have fixed opening costs of \$30, \$25 and \$35. Five customers with demands of 10, 15, 12, 8 and 10 units pay delivery costs in dollars per unit (rows customers, columns warehouses): [4,6,9], [5,4,7], [6,3,4], [8,5,3], [10,8,4]. Which warehouses should open, and who serves whom, to minimize the total cost in dollars?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 40 x_{11} + 60 x_{12} + 90 x_{13} + 75 x_{21} + 60 x_{22} + 105 x_{23} + 72 x_{31} + 36 x_{32} + 48 x_{33} + 64 x_{41} + 40 x_{42} + 24 x_{43} + 100 x_{51} + 80 x_{52} + 40 x_{53} + 30 y_{1} + 25 y_{2} + 35 y_{3} &&\text{make as small as possible: total cost: warehouse opening costs plus demand-weighted delivery costs (dollars)}\\ \text{s.t.}\;& x_{11} + x_{12} + x_{13} = 1 &&\text{limit 1: left side must equal 1}\\ \;& x_{21} + x_{22} + x_{23} = 1 &&\text{limit 2: left side must equal 1}\\ \;& x_{31} + x_{32} + x_{33} = 1 &&\text{limit 3: left side must equal 1}\\ \;& x_{41} + x_{42} + x_{43} = 1 &&\text{limit 4: left side must equal 1}\\ \;& x_{51} + x_{52} + x_{53} = 1 &&\text{limit 5: left side must equal 1}\\ \;& x_{11} \le y_{1} &&\text{limit 6: left side cannot exceed the right side}\\ \;& x_{21} \le y_{1} &&\text{limit 7: left side cannot exceed the right side}\\ \;& x_{31} \le y_{1} &&\text{limit 8: left side cannot exceed the right side}\\ \;& x_{41} \le y_{1} &&\text{limit 9: left side cannot exceed the right side}\\ \;& x_{51} \le y_{1} &&\text{limit 10: left side cannot exceed the right side}\\ \;& x_{12} \le y_{2} &&\text{limit 11: left side cannot exceed the right side}\\ \;& x_{22} \le y_{2} &&\text{limit 12: left side cannot exceed the right side}\\ \;& x_{32} \le y_{2} &&\text{limit 13: left side cannot exceed the right side}\\ \;& x_{42} \le y_{2} &&\text{limit 14: left side cannot exceed the right side}\\ \;& x_{52} \le y_{2} &&\text{limit 15: left side cannot exceed the right side}\\ \;& x_{13} \le y_{3} &&\text{limit 16: left side cannot exceed the right side}\\ \;& x_{23} \le y_{3} &&\text{limit 17: left side cannot exceed the right side}\\ \;& x_{33} \le y_{3} &&\text{limit 18: left side cannot exceed the right side}\\ \;& x_{43} \le y_{3} &&\text{limit 19: left side cannot exceed the right side}\\ \;& x_{53} \le y_{3} &&\text{limit 20: left side cannot exceed the right side}\\ \;& x_{11} \ge 0 &&\text{fraction of customer i's demand served from warehouse j (fraction) (i = 1, j = 1) cannot be negative}\\ \;& x_{12} \ge 0 &&\text{fraction of customer i's demand served from warehouse j &&\text{fraction of customer i's demand served from warehouse j (fraction) (i = 3, j = 1) cannot be negative}\\ \;& x_{32} \ge 0 &&\text{fraction of customer i's demand served from warehouse j ## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x11 = cp.Variable(name='x11')  # fraction of customer i's demand served from warehouse j (fraction) (i = 1, j = 1)
x12 = cp.Variable(name='x12')  # fraction of customer i's demand served from warehouse j (fraction) (i = 1, j = 2)
x13 = cp.Variable(name='x13')  # fraction of customer i's demand served from warehouse j (fraction) (i = 1, j = 3)
x21 = cp.Variable(name='x21')  # fraction of customer i's demand served from warehouse j (fraction) (i = 2, j = 1)
x22 = cp.Variable(name='x22')  # fraction of customer i's demand served from warehouse j (fraction) (i = 2, j = 2)
x23 = cp.Variable(name='x23')  # fraction of customer i's demand served from warehouse j (fraction) (i = 2, j = 3)
x31 = cp.Variable(name='x31')  # fraction of customer i's demand served from warehouse j (fraction) (i = 3, j = 1)
x32 = cp.Variable(name='x32')  # fraction of customer i's demand served from warehouse j (fraction) (i = 3, j = 2)
x33 = cp.Variable(name='x33')  # fraction of customer i's demand served from warehouse j (fraction) (i = 3, j = 3)
x41 = cp.Variable(name='x41')  # fraction of customer i's demand served from warehouse j (fraction) (i = 4, j = 1)
x42 = cp.Variable(name='x42')  # fraction of customer i's demand served from warehouse j (fraction) (i = 4, j = 2)
x43 = cp.Variable(name='x43')  # fraction of customer i's demand served from warehouse j (fraction) (i = 4, j = 3)
x51 = cp.Variable(name='x51')  # fraction of customer i's demand served from warehouse j (fraction) (i = 5, j = 1)
x52 = cp.Variable(name='x52')  # fraction of customer i's demand served from warehouse j (fraction) (i = 5, j = 2)
x53 = cp.Variable(name='x53')  # fraction of customer i's demand served from warehouse j (fraction) (i = 5, j = 3)
y1 = cp.Variable(name='y1', boolean=True)  # 1 if warehouse 1, 2, 3 is opened (yes/no)
y2 = cp.Variable(name='y2', boolean=True)  # 1 if warehouse 1, 2, 3 is opened (yes/no)
y3 = cp.Variable(name='y3', boolean=True)  # 1 if warehouse 1, 2, 3 is opened (yes/no)

objective = cp.Minimize(40*x11 + 60*x12 + 90*x13 + 75*x21 + 60*x22 + 105*x23 + 72*x31 + 36*x32 + 48*x33 + 64*x41 + 40*x42 + 24*x43 + 100*x51 + 80*x52 + 40*x53 + 30*y1 + 25*y2 + 35*y3)
constraints = [
    x11 + x12 + x13 == 1,
    x21 + x22 + x23 == 1,
    x31 + x32 + x33 == 1,
    x41 + x42 + x43 == 1,
    x51 + x52 + x53 == 1,
    x11 <= y1,
    x21 <= y1,
    x31 <= y1,
    x41 <= y1,
    x51 <= y1,
    x12 <= y2,
    x22 <= y2,
    x32 <= y2,
    x42 <= y2,
    x52 <= y2,
    x13 <= y3,
    x23 <= y3,
    x33 <= y3,
    x43 <= y3,
    x53 <= y3,
    x11 >= 0,
    x12 >= 0,
    x13 >= 0,
    x21 >= 0,
    x22 >= 0,
    x23 >= 0,
    x31 >= 0,
    x32 >= 0,
    x33 >= 0,
    x41 >= 0,
    x42 >= 0,
    x43 >= 0,
    x51 >= 0,
    x52 >= 0,
    x53 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x11 =', x11.value)
print('x12 =', x12.value)
print('x13 =', x13.value)
print('x21 =', x21.value)
print('x22 =', x22.value)
print('x23 =', x23.value)
print('x31 =', x31.value)
print('x32 =', x32.value)
print('x33 =', x33.value)
print('x41 =', x41.value)
print('x42 =', x42.value)
print('x43 =', x43.value)
print('x51 =', x51.value)
print('x52 =', x52.value)
print('x53 =', x53.value)
print('y1 =', y1.value)
print('y2 =', y2.value)
print('y3 =', y3.value)
```

All 35 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
