## The problem

A consumer with utility 0.3 ln x + 0.7 ln y has $100; good x costs $2 and good y costs $5. How much of each should she buy?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& 0.3 \log{\left(x \right)} + 0.7 \log{\left(y \right)} &&\text{make this as large as possible}\\ \text{s.t.}\;& 2 x + 5 y \le 100 &&\text{limit 1: left side cannot exceed 100}\\ \;& x \ge 0 &&\text{quantity of good x purchased (units) cannot be negative}\\ \;& y \ge 0 &&\text{quantity of good y purchased (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x = cp.Variable(name='x')  # quantity of good x purchased (units)
y = cp.Variable(name='y')  # quantity of good y purchased (units)

objective = cp.Maximize(0.3*cp.log(x) + 0.7*cp.log(y))
constraints = [
    2*x + 5*y <= 100,
    x >= 0,
    y >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('x =', x.value)
print('y =', y.value)
```

## The checked answer

You asked for: the optimal quantities of x and y to buy. The largest objective value is **2.65976**, at x = 15.002, y = 13.9992. All 3 constraints are satisfied (violation ≤0e+00). **Optimal, numerically checked.**

**Problem class:** Conic (convex); **solver:** CLARABEL.
