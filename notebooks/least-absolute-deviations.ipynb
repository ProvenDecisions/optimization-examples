## The problem

Fit y = a + b t to the points (0, 1), (1, 2), (2, 2), (3, 4), (4, 5) minimizing the sum of absolute residuals.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& \left|{a - 1}\right| + \left|{a + b - 2}\right| + \left|{a + 2 b - 2}\right| + \left|{a + 3 b - 4}\right| + \left|{a + 4 b - 5}\right| &&\text{make this as small as possible}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

b = cp.Variable(name='b')  # slope of fitted line (y per unit t)
a = cp.Variable(name='a')  # intercept of fitted line (y units)

objective = cp.Minimize(cp.abs(a - 1) + cp.abs(a + b - 2) + cp.abs(a + 2*b - 2) + cp.abs(a + 3*b - 4) + cp.abs(a + 4*b - 5))
constraints = [
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('b =', b.value)
print('a =', a.value)
```

## The checked answer

The smallest objective value is **1**, at b = 1, a = 1. **Optimal, numerically checked.**

**Problem class:** LP (after reformulating |.|, max, min); **solver:** HIGHS.
