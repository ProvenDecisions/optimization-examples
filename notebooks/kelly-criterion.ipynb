## The problem

A bet pays even money and wins with probability 0.6. What fraction of your bankroll should you stake each time to maximize long-run growth?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& 0.4 \log{\left(1 - f \right)} + 0.6 \log{\left(f + 1 \right)} &&\text{make this as large as possible}\\ \text{s.t.}\;& f \ge 0 &&\text{fraction of bankroll staked on each bet (fraction) cannot be negative}\\ \;& f \le 0.99 &&\text{fraction of bankroll staked on each bet (fraction) is at most 0.99}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

f = cp.Variable(name='f')  # fraction of bankroll staked on each bet (fraction)

objective = cp.Maximize(0.4*cp.log(1 - f) + 0.6*cp.log(f + 1))
constraints = [
    f >= 0,
    f <= 0.990000000000000,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('f =', f.value)
```

## The checked answer

You asked for: optimal stake fraction f. The largest objective value is **0.0201355**, at f = 0.200024. All 2 constraints are satisfied (violation ≤0e+00). **Optimal, numerically checked.**

**Problem class:** Conic (convex); **solver:** CLARABEL.
