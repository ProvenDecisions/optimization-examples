## The problem

A company makes products A and B. Each A needs 2 hours of labour and 1 unit of raw material and earns \$3; each B needs 1 hour of labour and 2 units of raw material and earns \$2. There are 100 labour hours and 80 units of raw material. Write the dual problem and find the dual prices of labour and raw material.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& 3 a + 2 b &&\text{make this as large as possible}\\ \text{s.t.}\;& 2 a + b \le 100 &&\text{limit 1: left side cannot exceed 100}\\ \;& a + 2 b \le 80 &&\text{limit 2: left side cannot exceed 80}\\ \;& a \ge 0 &&\text{units of product A produced (units) cannot be negative}\\ \;& b \ge 0 &&\text{units of product B produced (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

a = cp.Variable(name='a')  # units of product A produced (units)
b = cp.Variable(name='b')  # units of product B produced (units)

objective = cp.Maximize(3*a + 2*b)
constraints = [
    2*a + b <= 100,
    a + 2*b <= 80,
    a >= 0,
    b >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('a =', a.value)
print('b =', b.value)
```

## The checked answer

You asked for: the dual problem. Shadow prices (objective change per extra unit of each limit): (1) 2*a + b <= 100: **1.33333**; (2) a + 2*b <= 80: **0.333333**; (3) a >= 0: **0**; (4) b >= 0: **0**. The largest objective value is **160**, at a = 40, b = 20. All 4 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**
