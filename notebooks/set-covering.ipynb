## The problem

Six candidate stations (costs in $ millions) cover districts: S1 {1,2,3} cost 3; S2 {2,4} cost 2; S3 {3,4,5} cost 3; S4 {1,5} cost 2; S5 {4,5,6} cost 3; S6 {6} cost 1. Cover all districts 1-6 at minimum cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 3 s_{1} + 2 s_{2} + 3 s_{3} + 2 s_{4} + 3 s_{5} + s_{6} &&\text{make as small as possible: total cost of built stations (\$ millions)}\\ \text{s.t.}\;& s_{1} + s_{4} \ge 1 &&\text{limit 1: left side must reach at least 1}\\ \;& s_{1} + s_{2} \ge 1 &&\text{limit 2: left side must reach at least 1}\\ \;& s_{1} + s_{3} \ge 1 &&\text{limit 3: left side must reach at least 1}\\ \;& s_{2} + s_{3} + s_{5} \ge 1 &&\text{limit 4: left side must reach at least 1}\\ \;& s_{3} + s_{4} + s_{5} \ge 1 &&\text{limit 5: left side must reach at least 1}\\ \;& s_{5} + s_{6} \ge 1 &&\text{limit 6: left side must reach at least 1}\\ &s_{1}, s_{2}, s_{3}, s_{4}, s_{5}, s_{6} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

s1 = cp.Variable(name='s1', boolean=True)  # 1 if station S1 is built, 0 otherwise
s4 = cp.Variable(name='s4', boolean=True)  # 1 if station S4 is built, 0 otherwise
s2 = cp.Variable(name='s2', boolean=True)  # 1 if station S2 is built, 0 otherwise
s3 = cp.Variable(name='s3', boolean=True)  # 1 if station S3 is built, 0 otherwise
s5 = cp.Variable(name='s5', boolean=True)  # 1 if station S5 is built, 0 otherwise
s6 = cp.Variable(name='s6', boolean=True)  # 1 if station S6 is built, 0 otherwise

objective = cp.Minimize(3*s1 + 2*s2 + 3*s3 + 2*s4 + 3*s5 + s6)
constraints = [
    s1 + s4 >= 1,
    s1 + s2 >= 1,
    s1 + s3 >= 1,
    s2 + s3 + s5 >= 1,
    s3 + s4 + s5 >= 1,
    s5 + s6 >= 1,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('s1 =', s1.value)
print('s4 =', s4.value)
print('s2 =', s2.value)
print('s3 =', s3.value)
print('s5 =', s5.value)
print('s6 =', s6.value)
```

## The checked answer

The smallest objective value is **6**, at s1 = 1, s4 = 0, s2 = 0, s3 = 0, s5 = 1, s6 = 0. All 6 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**

**Problem class:** ILP; **solver:** HIGHS.
