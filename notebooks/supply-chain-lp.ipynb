## The problem

A factory makes up to 80 units at $5 each. Units go to warehouses W1 or W2 ($2 and $3 per unit), then to a store that needs 50 units ($4 from W1, $1 from W2). Each warehouse handles at most 40 units. Minimize total cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 f_{1} + 3 f_{2} + 5 p + 4 s_{1} + s_{2} &&\text{make this as small as possible}\\ \text{s.t.}\;& p \le 80 &&\text{units produced at the factory (units) is at most 80}\\ \;& f_{1} + f_{2} = p &&\text{limit 2: left side must equal the right side}\\ \;& s_{1} + s_{2} \ge 50 &&\text{limit 3: left side must reach at least 50}\\ \;& s_{1} = f_{1} &&\text{limit 4: left side must equal the right side}\\ \;& s_{2} = f_{2} &&\text{limit 5: left side must equal the right side}\\ \;& f_{1} \le 40 &&\text{units shipped from factory to warehouse W1 (units) is at most 40}\\ \;& f_{2} \le 40 &&\text{units shipped from factory to warehouse W2 (units) is at most 40}\\ \;& p \ge 0 &&\text{units produced at the factory (units) cannot be negative}\\ \;& f_{1} \ge 0 &&\text{units shipped from factory to warehouse W1 (units) cannot be negative}\\ \;& f_{2} \ge 0 &&\text{units shipped from factory to warehouse W2 (units) cannot be negative}\\ \;& s_{1} \ge 0 &&\text{units shipped from W1 to the store (units) cannot be negative}\\ \;& s_{2} \ge 0 &&\text{units shipped from W2 to the store (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

p = cp.Variable(name='p')  # units produced at the factory (units)
f1 = cp.Variable(name='f1')  # units shipped from factory to warehouse W1 (units)
f2 = cp.Variable(name='f2')  # units shipped from factory to warehouse W2 (units)
s1 = cp.Variable(name='s1')  # units shipped from W1 to the store (units)
s2 = cp.Variable(name='s2')  # units shipped from W2 to the store (units)

objective = cp.Minimize(2*f1 + 3*f2 + 5*p + 4*s1 + s2)
constraints = [
    p <= 80,
    f1 + f2 == p,
    s1 + s2 >= 50,
    s1 == f1,
    s2 == f2,
    f1 <= 40,
    f2 <= 40,
    p >= 0,
    f1 >= 0,
    f2 >= 0,
    s1 >= 0,
    s2 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('p =', p.value)
print('f1 =', f1.value)
print('f2 =', f2.value)
print('s1 =', s1.value)
print('s2 =', s2.value)
```

## The checked answer

The smallest objective value is **470**, at p = 50, f1 = 10, f2 = 40, s1 = 10, s2 = 40. All 12 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
