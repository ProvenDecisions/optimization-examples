## The problem

Three assets have expected returns 8%, 12% and 15%, volatilities 10%, 20% and 30%, and correlations 0.2 (1-2), 0.1 (1-3), 0.3 (2-3). Invest fully, no short sales, with expected return at least 11%. Minimize portfolio variance.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 0.01 w_{1}^{2} + 0.008 w_{1} \cdot w_{2} + 0.006 w_{1} \cdot w_{3} + 0.04 w_{2}^{2} + 0.036 w_{2} \cdot w_{3} + 0.09 w_{3}^{2} &&\text{make this as small as possible}\\ \text{s.t.}\;& w_{1} + w_{2} + w_{3} = 1 &&\text{limit 1: left side must equal 1}\\ \;& 0.08 w_{1} + 0.12 w_{2} + 0.15 w_{3} \ge 0.11 &&\text{limit 2: left side must reach at least 0.11}\\ \;& w_{1} \ge 0 &&\text{fraction of portfolio invested in asset 1 (fraction) cannot be negative}\\ \;& w_{2} \ge 0 &&\text{fraction of portfolio invested in asset 2 (fraction) cannot be negative}\\ \;& w_{3} \ge 0 &&\text{fraction of portfolio invested in asset 3 (fraction) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp
import numpy as np

w1 = cp.Variable(name='w1')  # fraction of portfolio invested in asset 1 (fraction)
w2 = cp.Variable(name='w2')  # fraction of portfolio invested in asset 2 (fraction)
w3 = cp.Variable(name='w3')  # fraction of portfolio invested in asset 3 (fraction)

objective = cp.Minimize(cp.quad_form(cp.hstack([w1, w2, w3]), np.array([[0.01, 0.004, 0.003], [0.004, 0.04, 0.018], [0.003, 0.018, 0.09]])))
constraints = [
    w1 + w2 + w3 == 1,
    0.08*w1 + 0.12*w2 + 0.15*w3 >= 0.11,
    w1 >= 0,
    w2 >= 0,
    w3 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('w1 =', w1.value)
print('w2 =', w2.value)
print('w3 =', w3.value)
```

## The checked answer

The smallest objective value is **0.0159946**, at w1 = 0.430428, w2 = 0.329002, w3 = 0.24057. All 5 constraints are satisfied (largest violation 1e-16, within the solver's tolerance). **Optimal, numerically checked.**
