## The problem

An order of at least 12 parts can be made on two machines. Machine 1 costs \$20 to set up, then \$3 per part, and makes at most 10 parts; machine 2 costs \$8 to set up, then \$6 per part, and makes at most 15 parts. A machine that makes no parts is not set up. Parts are whole numbers. Minimize total cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 3 x_{1} + 6 x_{2} + 20 y_{1} + 8 y_{2} &&\text{make as small as possible: total cost of setup plus per-part costs (\$)}\\ \text{s.t.}\;& x_{1} + x_{2} \ge 12 &&\text{limit 1: left side must reach at least 12}\\ \;& x_{1} \le 10 y_{1} &&\text{limit 2: left side cannot exceed the right side}\\ \;& x_{2} \le 15 y_{2} &&\text{limit 3: left side cannot exceed the right side}\\ \;& x_{1} \ge 0 &&\text{parts made on machine 1 (parts) cannot be negative}\\ \;& x_{2} \ge 0 &&\text{parts made on machine 2 (parts) cannot be negative}\\ &x_{1}, x_{2} \in \mathbb{Z} &&\text{whole numbers only}\\ &y_{1}, y_{2} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1', integer=True)  # parts made on machine 1 (parts)
x2 = cp.Variable(name='x2', integer=True)  # parts made on machine 2 (parts)
y1 = cp.Variable(name='y1', boolean=True)  # 1 if machine 1 is set up, 0 otherwise (yes/no)
y2 = cp.Variable(name='y2', boolean=True)  # 1 if machine 2 is set up, 0 otherwise (yes/no)

objective = cp.Minimize(3*x1 + 6*x2 + 20*y1 + 8*y2)
constraints = [
    x1 + x2 >= 12,
    x1 <= 10*y1,
    x2 <= 15*y2,
    x1 >= 0,
    x2 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('x2 =', x2.value)
print('y1 =', y1.value)
print('y2 =', y2.value)
```

## The checked answer

The smallest objective value is **70**, at x1 = 10, x2 = 2, y1 = 1, y2 = 1. All 5 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
