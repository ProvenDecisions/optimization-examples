## The problem

Road lengths in km: s-a 1, s-b 4, a-b 2, a-t 6, b-t 1 (one-way, in that direction). Find the shortest route from s to t.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 x_{ab} + 6 x_{at} + x_{bt} + x_{sa} + 4 x_{sb} &&\text{make as small as possible: total route length from s to t (km)}\\ \text{s.t.}\;& x_{sa} + x_{sb} = 1 &&\text{limit 1: left side must equal 1}\\ \;& - x_{ab} - x_{at} + x_{sa} = 0 &&\text{limit 2: left side must equal 0}\\ \;& x_{ab} - x_{bt} + x_{sb} = 0 &&\text{limit 3: left side must equal 0}\\ \;& x_{at} + x_{bt} = 1 &&\text{limit 4: left side must equal 1}\\ \;& x_{sa} \ge 0 &&\text{flow on road s-a (1 if used) cannot be negative}\\ \;& x_{sb} \ge 0 &&\text{flow on road s-b (1 if used) cannot be negative}\\ \;& x_{ab} \ge 0 &&\text{flow on road a-b (1 if used) cannot be negative}\\ \;& x_{at} \ge 0 &&\text{flow on road a-t (1 if used) cannot be negative}\\ \;& x_{bt} \ge 0 &&\text{flow on road b-t (1 if used) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x_sa = cp.Variable(name='x_sa')  # flow on road s-a (1 if used)
x_sb = cp.Variable(name='x_sb')  # flow on road s-b (1 if used)
x_ab = cp.Variable(name='x_ab')  # flow on road a-b (1 if used)
x_at = cp.Variable(name='x_at')  # flow on road a-t (1 if used)
x_bt = cp.Variable(name='x_bt')  # flow on road b-t (1 if used)

objective = cp.Minimize(2*x_ab + 6*x_at + x_bt + x_sa + 4*x_sb)
constraints = [
    x_sa + x_sb == 1,
    -x_ab - x_at + x_sa == 0,
    x_ab - x_bt + x_sb == 0,
    x_at + x_bt == 1,
    x_sa >= 0,
    x_sb >= 0,
    x_ab >= 0,
    x_at >= 0,
    x_bt >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x_sa =', x_sa.value)
print('x_sb =', x_sb.value)
print('x_ab =', x_ab.value)
print('x_at =', x_at.value)
print('x_bt =', x_bt.value)
```

## The checked answer

The smallest objective value is **4**, at x_sa = 1, x_sb = 0, x_ab = 1, x_at = 0, x_bt = 1. All 9 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
