## The problem

Classify points on a line: 2, 3 and 4 are class +1; 0, 1 and 2.5 are class -1. Find w and b minimizing 0.5 w^2 + C * (sum of hinge losses max(0, 1 - y_i (w x_i + b))) with C = 1.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& h_{1} + h_{2} + h_{3} + h_{4} + h_{5} + h_{6} + 0.5 w^{2} &&\text{make this as small as possible}\\ \text{s.t.}\;& h_{1} \ge - b - 2 w + 1 &&\text{limit 1: left side must reach at least the right side}\\ \;& h_{2} \ge - b - 3 w + 1 &&\text{limit 2: left side must reach at least the right side}\\ \;& h_{3} \ge - b - 4 w + 1 &&\text{limit 3: left side must reach at least the right side}\\ \;& h_{4} \ge b + 1 &&\text{limit 4: left side must reach at least the right side}\\ \;& h_{5} \ge b + w + 1 &&\text{limit 5: left side must reach at least the right side}\\ \;& h_{6} \ge b + 2.5 w + 1 &&\text{limit 6: left side must reach at least the right side}\\ \;& h_{1} \ge 0 &&\text{hinge loss for point x=2, class +1 cannot be negative}\\ \;& h_{2} \ge 0 &&\text{hinge loss for point x=3, class +1 cannot be negative}\\ \;& h_{3} \ge 0 &&\text{hinge loss for point x=4, class +1 cannot be negative}\\ \;& h_{4} \ge 0 &&\text{hinge loss for point x=0, class -1 cannot be negative}\\ \;& h_{5} \ge 0 &&\text{hinge loss for point x=1, class -1 cannot be negative}\\ \;& h_{6} \ge 0 &&\text{hinge loss for point x=2.5, class -1 cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

h1 = cp.Variable(name='h1')  # hinge loss for point x=2, class +1
w = cp.Variable(name='w')  # slope of the separating line (per unit x)
b = cp.Variable(name='b')  # intercept/bias of the separating line
h2 = cp.Variable(name='h2')  # hinge loss for point x=3, class +1
h3 = cp.Variable(name='h3')  # hinge loss for point x=4, class +1
h4 = cp.Variable(name='h4')  # hinge loss for point x=0, class -1
h5 = cp.Variable(name='h5')  # hinge loss for point x=1, class -1
h6 = cp.Variable(name='h6')  # hinge loss for point x=2.5, class -1

objective = cp.Minimize(h1 + h2 + h3 + h4 + h5 + h6 + 0.5*w**2)
constraints = [
    h1 >= -b - 2*w + 1,
    h2 >= -b - 3*w + 1,
    h3 >= -b - 4*w + 1,
    h4 >= b + 1,
    h5 >= b + w + 1,
    h6 >= b + 2.5*w + 1,
    h1 >= 0,
    h2 >= 0,
    h3 >= 0,
    h4 >= 0,
    h5 >= 0,
    h6 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('h1 =', h1.value)
print('w =', w.value)
print('b =', b.value)
print('h2 =', h2.value)
print('h3 =', h3.value)
print('h4 =', h4.value)
print('h5 =', h5.value)
print('h6 =', h6.value)
```

## The checked answer

The smallest objective value is **3**, at h1 = 1, w = 1, b = -2, h2 = 9.95069e-09, h3 = 0, h4 = 0, h5 = 9.4156e-09, h6 = 1.5. All 12 constraints are satisfied (violation ≤2e-11). **Optimal, numerically checked.**
