## The problem

Oats, milk and beans cost $0.30, $0.25 and $0.50 per serving. Per serving, protein (grams): 5, 8, 9; calcium (% of daily value): 1, 30, 4; iron (% of daily value): 4, 0, 3. A day needs at least 40 grams of protein, 60% of the daily value of calcium and 20% of the daily value of iron. Find the cheapest diet.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 0.5\,\mathrm{beans} + 0.25\,\mathrm{milk} + 0.3\,\mathrm{oats} &&\text{make as small as possible: total daily diet cost (dollars)}\\ \text{s.t.}\;& 9\,\mathrm{beans} + 8\,\mathrm{milk} + 5\,\mathrm{oats} \ge 40 &&\text{limit 1: left side must reach at least 40}\\ \;& 4\,\mathrm{beans} + 30\,\mathrm{milk} + \mathrm{oats} \ge 60 &&\text{limit 2: left side must reach at least 60}\\ \;& 3\,\mathrm{beans} + 4\,\mathrm{oats} \ge 20 &&\text{limit 3: left side must reach at least 20}\\ \;& \mathrm{oats} \ge 0 &&\text{servings of oats per day (servings) cannot be negative}\\ \;& \mathrm{milk} \ge 0 &&\text{servings of milk per day (servings) cannot be negative}\\ \;& \mathrm{beans} \ge 0 &&\text{servings of beans per day (servings) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

oats = cp.Variable(name='oats')  # servings of oats per day (servings)
milk = cp.Variable(name='milk')  # servings of milk per day (servings)
beans = cp.Variable(name='beans')  # servings of beans per day (servings)

objective = cp.Minimize(0.5*beans + 0.25*milk + 0.3*oats)
constraints = [
    9*beans + 8*milk + 5*oats >= 40,
    4*beans + 30*milk + oats >= 60,
    3*beans + 4*oats >= 20,
    oats >= 0,
    milk >= 0,
    beans >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('oats =', oats.value)
print('milk =', milk.value)
print('beans =', beans.value)
```

## The checked answer

Assumed: servings may be fractional. The smallest objective value is **1.96875**, at oats = 5, milk = 1.875, beans = 0. All 6 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
