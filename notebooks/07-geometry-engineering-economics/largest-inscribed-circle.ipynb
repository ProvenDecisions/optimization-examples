## The problem

Find the largest circle that fits inside the region 0 <= x <= 2, 0 <= y <= 2, x + y <= 3. Report its radius.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& r &&\text{make this as large as possible}\\ \text{s.t.}\;& a - r \ge 0 &&\text{limit 1: left side must reach at least 0}\\ \;& a + r \le 2 &&\text{limit 2: left side cannot exceed 2}\\ \;& b - r \ge 0 &&\text{limit 3: left side must reach at least 0}\\ \;& b + r \le 2 &&\text{limit 4: left side cannot exceed 2}\\ \;& a + b + \sqrt{2} \cdot r \le 3 &&\text{limit 5: left side cannot exceed 3}\\ \;& r \ge 0 &&\text{radius of circle cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

a = cp.Variable(name='a')  # x-coordinate of circle center
r = cp.Variable(name='r')  # radius of circle
b = cp.Variable(name='b')  # y-coordinate of circle center

objective = cp.Maximize(r)
constraints = [
    a - r >= 0,
    a + r <= 2,
    b - r >= 0,
    b + r <= 2,
    a + b + 1.4142135623730951*r <= 3,
    r >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('a =', a.value)
print('r =', r.value)
print('b =', b.value)
```

## The checked answer

You asked for: radius r of the largest inscribed circle. The largest objective value is **0.87868**, at a = 0.87868, r = 0.87868, b = 0.87868. All 6 constraints are satisfied (violation ≤4e-16). **Optimal, numerically checked.**
