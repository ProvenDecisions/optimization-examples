## The problem

Towns sit at (0, 0), (4, 0) and (1, 3) on a map in km. Where should a warehouse go to minimize the total straight-line distance to the three towns, and what is that total?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& \sqrt{x^{2} + y^{2}} + \sqrt{y^{2} + \left(x - 4\right)^{2}} + \sqrt{\left(x - 1\right)^{2} + \left(y - 3\right)^{2}} &&\text{make as small as possible: total straight-line distance from the warehouse to the three towns (km)}\\ \text{s.t.}\;& x \ge -1000 &&\text{warehouse east-west coordinate (km) is at least -1000}\\ \;& y \ge -1000 &&\text{warehouse north-south coordinate (km) is at least -1000}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x = cp.Variable(name='x')  # warehouse east-west coordinate (km)
y = cp.Variable(name='y')  # warehouse north-south coordinate (km)

objective = cp.Minimize(cp.norm(cp.hstack([x, y]), 2) + cp.norm(cp.hstack([y, x - 4]), 2) + cp.norm(cp.hstack([x - 1, y - 3]), 2))
constraints = [
    x >= -1000,
    y >= -1000,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('x =', x.value)
print('y =', y.value)
```

## The checked answer

Assumed: the warehouse may be placed anywhere on the plane. The smallest objective value is **6.54099**, at x = 1.30224, y = 1.04678. All 2 constraints are satisfied exactly. **Optimal, numerically checked.**
