## The problem

Two features have covariance matrix [[4, 2], [2, 3]]. Find the direction (unit vector) of greatest variance and the variance along it.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& 4 w_{1}^{2} + 4 w_{1} \cdot w_{2} + 3 w_{2}^{2} &&\text{make this as large as possible}\\ \text{s.t.}\;& w_{1}^{2} + w_{2}^{2} = 1 &&\text{limit 1: left side must equal 1}\end{aligned}
$$

## Solve it with NumPy

Run the cell: the answer is an eigenvalue, so no optimization solver is needed.

```python
import numpy as np

# maximize x'Ax over x'x = 1: the largest eigenvalue (spectral theorem)
A = np.array([[4, 2], [2, 3]])  # the quadratic form, x = (w1, w2)
vals, vecs = np.linalg.eigh(A)
x = np.sqrt(1) * vecs[:, -1]
x = x if x[np.argmax(np.abs(x))] > 0 else -x  # the sign is arbitrary; show the largest entry positive
print('optimal', 1 * vals[-1])
for name, v in zip(['w1', 'w2'], x):
    print(name, '=', v)
```

## The checked answer

You asked for: the unit direction (w1, w2) of greatest variance and the variance along it (the objective value). The largest objective value is **5.56155**, at w1 = 0.788205, w2 = 0.615412. The constraint is satisfied (largest violation 1e-16, within the solver's tolerance). **Optimal, numerically checked.**

**Problem class:** Eigenvalue problem; **solver:** eigenvalue decomposition (NumPy).
