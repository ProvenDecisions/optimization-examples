## The problem

Arcs and capacities (units per hour): s->a 10, s->b 5, a->b 15, a->t 5, b->t 10. What is the maximum flow from s to t, in units per hour?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& f &&\text{make as large as possible: total flow from s to t (units per hour)}\\ \text{s.t.}\;& f = f_{sa} + f_{sb} &&\text{limit 1: left side must equal the right side}\\ \;& f = f_{at} + f_{bt} &&\text{limit 2: left side must equal the right side}\\ \;& f_{sa} = f_{ab} + f_{at} &&\text{limit 3: left side must equal the right side}\\ \;& f_{ab} + f_{sb} = f_{bt} &&\text{limit 4: left side must equal the right side}\\ \;& f_{sa} \le 10 &&\text{flow on arc s->a (units per hour) is at most 10}\\ \;& f_{sb} \le 5 &&\text{flow on arc s->b (units per hour) is at most 5}\\ \;& f_{ab} \le 15 &&\text{flow on arc a->b (units per hour) is at most 15}\\ \;& f_{at} \le 5 &&\text{flow on arc a->t (units per hour) is at most 5}\\ \;& f_{bt} \le 10 &&\text{flow on arc b->t (units per hour) is at most 10}\\ \;& f \ge 0 &&\text{total flow from s to t (units per hour) cannot be negative}\\ \;& f_{sa} \ge 0 &&\text{flow on arc s->a (units per hour) cannot be negative}\\ \;& f_{sb} \ge 0 &&\text{flow on arc s->b (units per hour) cannot be negative}\\ \;& f_{ab} \ge 0 &&\text{flow on arc a->b (units per hour) cannot be negative}\\ \;& f_{at} \ge 0 &&\text{flow on arc a->t (units per hour) cannot be negative}\\ \;& f_{bt} \ge 0 &&\text{flow on arc b->t (units per hour) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

f = cp.Variable(name='f')  # total flow from s to t (units per hour)
f_sa = cp.Variable(name='f_sa')  # flow on arc s->a (units per hour)
f_sb = cp.Variable(name='f_sb')  # flow on arc s->b (units per hour)
f_at = cp.Variable(name='f_at')  # flow on arc a->t (units per hour)
f_bt = cp.Variable(name='f_bt')  # flow on arc b->t (units per hour)
f_ab = cp.Variable(name='f_ab')  # flow on arc a->b (units per hour)

objective = cp.Maximize(f)
constraints = [
    f == f_sa + f_sb,
    f == f_at + f_bt,
    f_sa == f_ab + f_at,
    f_ab + f_sb == f_bt,
    f_sa <= 10,
    f_sb <= 5,
    f_ab <= 15,
    f_at <= 5,
    f_bt <= 10,
    f >= 0,
    f_sa >= 0,
    f_sb >= 0,
    f_ab >= 0,
    f_at >= 0,
    f_bt >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('f =', f.value)
print('f_sa =', f_sa.value)
print('f_sb =', f_sb.value)
print('f_at =', f_at.value)
print('f_bt =', f_bt.value)
print('f_ab =', f_ab.value)
```

## The checked answer

The largest objective value is **15**, at f = 15, f_sa = 10, f_sb = 5, f_at = 5, f_bt = 10, f_ab = 5. All 15 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**
