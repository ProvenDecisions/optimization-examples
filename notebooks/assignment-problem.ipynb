## The problem

Workers A, B, C take these hours on jobs 1, 2, 3: A = 9, 2, 7; B = 6, 4, 3; C = 5, 8, 1. Each worker does exactly one job and each job gets one worker. Minimize total hours.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 9 x_{A1} + 2 x_{A2} + 7 x_{A3} + 6 x_{B1} + 4 x_{B2} + 3 x_{B3} + 5 x_{C1} + 8 x_{C2} + x_{C3} &&\text{make as small as possible: total hours worked (hours)}\\ \text{s.t.}\;& x_{A1} + x_{A2} + x_{A3} = 1 &&\text{limit 1: left side must equal 1}\\ \;& x_{B1} + x_{B2} + x_{B3} = 1 &&\text{limit 2: left side must equal 1}\\ \;& x_{C1} + x_{C2} + x_{C3} = 1 &&\text{limit 3: left side must equal 1}\\ \;& x_{A1} + x_{B1} + x_{C1} = 1 &&\text{limit 4: left side must equal 1}\\ \;& x_{A2} + x_{B2} + x_{C2} = 1 &&\text{limit 5: left side must equal 1}\\ \;& x_{A3} + x_{B3} + x_{C3} = 1 &&\text{limit 6: left side must equal 1}\\ &x_{A1}, x_{A2}, x_{A3}, x_{B1}, x_{B2}, x_{B3}, x_{C1}, x_{C2}, x_{C3} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

xA1 = cp.Variable(name='xA1', boolean=True)  # 1 if worker A does job 1, else 0
xA2 = cp.Variable(name='xA2', boolean=True)  # 1 if worker A does job 2, else 0
xA3 = cp.Variable(name='xA3', boolean=True)  # 1 if worker A does job 3, else 0
xB1 = cp.Variable(name='xB1', boolean=True)  # 1 if worker B does job 1, else 0
xB2 = cp.Variable(name='xB2', boolean=True)  # 1 if worker B does job 2, else 0
xB3 = cp.Variable(name='xB3', boolean=True)  # 1 if worker B does job 3, else 0
xC1 = cp.Variable(name='xC1', boolean=True)  # 1 if worker C does job 1, else 0
xC2 = cp.Variable(name='xC2', boolean=True)  # 1 if worker C does job 2, else 0
xC3 = cp.Variable(name='xC3', boolean=True)  # 1 if worker C does job 3, else 0

objective = cp.Minimize(9*xA1 + 2*xA2 + 7*xA3 + 6*xB1 + 4*xB2 + 3*xB3 + 5*xC1 + 8*xC2 + xC3)
constraints = [
    xA1 + xA2 + xA3 == 1,
    xB1 + xB2 + xB3 == 1,
    xC1 + xC2 + xC3 == 1,
    xA1 + xB1 + xC1 == 1,
    xA2 + xB2 + xC2 == 1,
    xA3 + xB3 + xC3 == 1,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('xA1 =', xA1.value)
print('xA2 =', xA2.value)
print('xA3 =', xA3.value)
print('xB1 =', xB1.value)
print('xB2 =', xB2.value)
print('xB3 =', xB3.value)
print('xC1 =', xC1.value)
print('xC2 =', xC2.value)
print('xC3 =', xC3.value)
```

## The checked answer

The smallest objective value is **9**, at xA1 = 0, xA2 = 1, xA3 = 0, xB1 = 1, xB2 = 0, xB3 = 0, xC1 = 0, xC2 = 0, xC3 = 1. All 6 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**
