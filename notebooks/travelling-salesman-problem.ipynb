## The problem

Distances in km between cities 1-5 (symmetric): d12=3, d13=4, d14=2, d15=7, d23=4, d24=6, d25=3, d34=5, d35=8, d45=6. Find the shortest tour (in km) visiting every city once and returning to the start.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 3 y_{12} + 4 y_{13} + 2 y_{14} + 7 y_{15} + 4 y_{23} + 6 y_{24} + 3 y_{25} + 5 y_{34} + 8 y_{35} + 6 y_{45} &&\text{make as small as possible: total tour length (km)}\\ \text{s.t.}\;& y_{12} + y_{13} + y_{14} + y_{15} = 2 &&\text{limit 1: left side must equal 2}\\ \;& y_{12} + y_{23} + y_{24} + y_{25} = 2 &&\text{limit 2: left side must equal 2}\\ \;& y_{13} + y_{23} + y_{34} + y_{35} = 2 &&\text{limit 3: left side must equal 2}\\ \;& y_{14} + y_{24} + y_{34} + y_{45} = 2 &&\text{limit 4: left side must equal 2}\\ \;& y_{15} + y_{25} + y_{35} + y_{45} = 2 &&\text{limit 5: left side must equal 2}\\ &y_{12}, y_{13}, y_{14}, y_{15}, y_{23}, y_{24}, y_{25}, y_{34}, y_{35}, y_{45} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

y12 = cp.Variable(name='y12', boolean=True)  # 1 if the tour uses the road between cities 1 and 2 (else 0); likewise y13, y14, y15, y23, y24, y25, y34, y35, y45
y13 = cp.Variable(name='y13', boolean=True)  # see y12
y14 = cp.Variable(name='y14', boolean=True)  # see y12
y15 = cp.Variable(name='y15', boolean=True)  # see y12
y23 = cp.Variable(name='y23', boolean=True)  # see y12
y24 = cp.Variable(name='y24', boolean=True)  # see y12
y25 = cp.Variable(name='y25', boolean=True)  # see y12
y34 = cp.Variable(name='y34', boolean=True)  # see y12
y35 = cp.Variable(name='y35', boolean=True)  # see y12
y45 = cp.Variable(name='y45', boolean=True)  # see y12

objective = cp.Minimize(3*y12 + 4*y13 + 2*y14 + 7*y15 + 4*y23 + 6*y24 + 3*y25 + 5*y34 + 8*y35 + 6*y45)
constraints = [
    y12 + y13 + y14 + y15 == 2,
    y12 + y23 + y24 + y25 == 2,
    y13 + y23 + y34 + y35 == 2,
    y14 + y24 + y34 + y45 == 2,
    y15 + y25 + y35 + y45 == 2,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('y12 =', y12.value)
print('y13 =', y13.value)
print('y14 =', y14.value)
print('y15 =', y15.value)
print('y23 =', y23.value)
print('y24 =', y24.value)
print('y25 =', y25.value)
print('y34 =', y34.value)
print('y35 =', y35.value)
print('y45 =', y45.value)
```

## The checked answer

Assumed: the tour is undirected, so each city has exactly two tour edges; with 5 cities this already rules out subtours, since a 3-city loop plus a 2-city loop would need an edge used twice. The smallest objective value is **19**, at y12 = 0, y13 = 1, y14 = 1, y15 = 0, y23 = 1, y24 = 0, y25 = 1, y34 = 0, y35 = 0, y45 = 1. All 5 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**

**Problem class:** ILP; **solver:** HIGHS.
