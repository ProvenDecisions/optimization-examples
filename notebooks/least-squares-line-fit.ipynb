## The problem

minimize (m + b - 1)^2 + (2m + b - 3)^2 + (3m + b - 4)^2 + (4m + b - 4)^2

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& \left(b + m - 1\right)^{2} + \left(b + 2 m - 3\right)^{2} + \left(b + 3 m - 4\right)^{2} + \left(b + 4 m - 4\right)^{2} &&\text{make as small as possible: sum of squared residuals of the line y = m*x + b against the points (1,1), (2,3), (3,4), (4,4)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

b = cp.Variable(name='b')  # intercept of the fitted line (unitless)
m = cp.Variable(name='m')  # slope of the fitted line (unitless)

objective = cp.Minimize((b + m - 1)**2 + (b + 2*m - 3)**2 + (b + 3*m - 4)**2 + (b + 4*m - 4)**2)
constraints = [
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('b =', b.value)
print('m =', m.value)
```

## The checked answer

The smallest objective value is **1**, at b = 0.5, m = 1. **Optimal, numerically checked.**
