## The problem

minimize 2x + y + 3z
subject to
  x + y + z = 10
  x - y >= 2
  x, y, z >= 0
# asked: the dual problem and its optimal dual prices

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 x + y + 3 z &&\text{make as small as possible: total cost 2x + y + 3z (cost units)}\\ \text{s.t.}\;& x + y + z = 10 &&\text{limit 1: left side must equal 10}\\ \;& x - y \ge 2 &&\text{limit 2: left side must reach at least 2}\\ \;& x \ge 0 &&\text{decision variable x in the primal (units) cannot be negative}\\ \;& y \ge 0 &&\text{decision variable y in the primal (units) cannot be negative}\\ \;& z \ge 0 &&\text{decision variable z in the primal (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x = cp.Variable(name='x')  # decision variable x in the primal (units)
y = cp.Variable(name='y')  # decision variable y in the primal (units)
z = cp.Variable(name='z')  # decision variable z in the primal (units)

objective = cp.Minimize(2*x + y + 3*z)
constraints = [
    x + y + z == 10,
    x - y >= 2,
    x >= 0,
    y >= 0,
    z >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x =', x.value)
print('y =', y.value)
print('z =', z.value)
```

## The checked answer

You asked for: the dual problem and its optimal dual prices. Shadow prices (objective change per extra unit of each limit): (1) x + y + z = 10: **1.5**; (2) x - y >= 2: **0.5**; (3) x >= 0: **0**; (4) y >= 0: **0**; (5) z >= 0: **1.5**. The smallest objective value is **16**, at x = 6, y = 4, z = 0. All 5 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**
