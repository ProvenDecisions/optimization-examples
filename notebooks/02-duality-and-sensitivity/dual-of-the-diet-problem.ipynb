## The problem

Three foods cost \$3, \$4 and \$5 per serving. They supply 2, 1 and 1 grams of protein and 1, 2 and 3 grams of fibre per serving. A diet needs at least 8 grams of protein and 10 grams of fibre at minimum cost. Write the dual problem and find the price of each nutrient.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 3 x_{1} + 4 x_{2} + 5 x_{3} &&\text{make as small as possible: total diet cost (dollars)}\\ \text{s.t.}\;& 2 x_{1} + x_{2} + x_{3} \ge 8 &&\text{limit 1: left side must reach at least 8}\\ \;& x_{1} + 2 x_{2} + 3 x_{3} \ge 10 &&\text{limit 2: left side must reach at least 10}\\ \;& x_{1} \ge 0 &&\text{servings of food 1 (servings) cannot be negative}\\ \;& x_{2} \ge 0 &&\text{servings of food 2 (servings) cannot be negative}\\ \;& x_{3} \ge 0 &&\text{servings of food 3 (servings) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1')  # servings of food 1 (servings)
x2 = cp.Variable(name='x2')  # servings of food 2 (servings)
x3 = cp.Variable(name='x3')  # servings of food 3 (servings)

objective = cp.Minimize(3*x1 + 4*x2 + 5*x3)
constraints = [
    2*x1 + x2 + x3 >= 8,
    x1 + 2*x2 + 3*x3 >= 10,
    x1 >= 0,
    x2 >= 0,
    x3 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('x2 =', x2.value)
print('x3 =', x3.value)
```

## The checked answer

You asked for: the dual problem and the price of each nutrient (dual values of the protein and fibre constraints). Shadow prices (objective change per extra unit of each limit): (1) 2*x1 + x2 + x3 >= 8: **0.8**; (2) x1 + 2*x2 + 3*x3 >= 10: **1.4**; (3) x1 >= 0: **0**; (4) x2 >= 0: **0.4**; (5) x3 >= 0: **0**. The smallest objective value is **20.4**, at x1 = 2.8, x2 = 0, x3 = 2.4. All 5 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**
