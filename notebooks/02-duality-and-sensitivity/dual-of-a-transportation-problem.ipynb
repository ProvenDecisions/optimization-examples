## The problem

Two warehouses hold 30 and 25 units. Two shops need exactly 20 and 30 units. Shipping one unit costs \$4 and \$6 from warehouse 1 to shops 1 and 2, and \$5 and \$3 from warehouse 2. Minimize shipping cost. Write the dual problem and find the dual prices of supply and demand.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 4 x_{11} + 6 x_{12} + 5 x_{21} + 3 x_{22} &&\text{make as small as possible: total shipping cost (dollars)}\\ \text{s.t.}\;& x_{11} + x_{12} \le 30 &&\text{limit 1: left side cannot exceed 30}\\ \;& x_{21} + x_{22} \le 25 &&\text{limit 2: left side cannot exceed 25}\\ \;& x_{11} + x_{21} = 20 &&\text{limit 3: left side must equal 20}\\ \;& x_{12} + x_{22} = 30 &&\text{limit 4: left side must equal 30}\\ \;& x_{11} \ge 0 &&\text{units shipped from warehouse 1 to shop 1 (units) cannot be negative}\\ \;& x_{12} \ge 0 &&\text{units shipped from warehouse 1 to shop 2 (units) cannot be negative}\\ \;& x_{21} \ge 0 &&\text{units shipped from warehouse 2 to shop 1 (units) cannot be negative}\\ \;& x_{22} \ge 0 &&\text{units shipped from warehouse 2 to shop 2 (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x11 = cp.Variable(name='x11')  # units shipped from warehouse 1 to shop 1 (units)
x12 = cp.Variable(name='x12')  # units shipped from warehouse 1 to shop 2 (units)
x21 = cp.Variable(name='x21')  # units shipped from warehouse 2 to shop 1 (units)
x22 = cp.Variable(name='x22')  # units shipped from warehouse 2 to shop 2 (units)

objective = cp.Minimize(4*x11 + 6*x12 + 5*x21 + 3*x22)
constraints = [
    x11 + x12 <= 30,
    x21 + x22 <= 25,
    x11 + x21 == 20,
    x12 + x22 == 30,
    x11 >= 0,
    x12 >= 0,
    x21 >= 0,
    x22 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x11 =', x11.value)
print('x12 =', x12.value)
print('x21 =', x21.value)
print('x22 =', x22.value)
```

Assumed: warehouse supplies are upper limits, so not all stock must be shipped. The smallest objective value is **185**, at x11 = 20, x12 = 5, x21 = 0, x22 = 25.
