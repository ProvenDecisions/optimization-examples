## The problem

Demand for the next three months is 40, 60 and 50 units. Production costs \$10, \$12 and \$11 per unit in months 1-3, capacity is 55 units a month, and holding a unit to the next month costs \$1. Start and end with no inventory. Minimize total cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& i_{1} + i_{2} + 10 p_{1} + 12 p_{2} + 11 p_{3} &&\text{make this as small as possible}\\ \text{s.t.}\;& - i_{1} + p_{1} = 40 &&\text{limit 1: left side must equal 40}\\ \;& i_{1} - i_{2} + p_{2} = 60 &&\text{limit 2: left side must equal 60}\\ \;& i_{2} + p_{3} = 50 &&\text{limit 3: left side must equal 50}\\ \;& p_{1} \le 55 &&\text{units produced in month 1 (units) is at most 55}\\ \;& p_{2} \le 55 &&\text{units produced in month 2 (units) is at most 55}\\ \;& p_{3} \le 55 &&\text{units produced in month 3 (units) is at most 55}\\ \;& p_{1} \ge 0 &&\text{units produced in month 1 (units) cannot be negative}\\ \;& p_{2} \ge 0 &&\text{units produced in month 2 (units) cannot be negative}\\ \;& p_{3} \ge 0 &&\text{units produced in month 3 (units) cannot be negative}\\ \;& i_{1} \ge 0 &&\text{inventory held from month 1 to month 2 (units) cannot be negative}\\ \;& i_{2} \ge 0 &&\text{inventory held from month 2 to month 3 (units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

p1 = cp.Variable(name='p1')  # units produced in month 1 (units)
i1 = cp.Variable(name='i1')  # inventory held from month 1 to month 2 (units)
p2 = cp.Variable(name='p2')  # units produced in month 2 (units)
i2 = cp.Variable(name='i2')  # inventory held from month 2 to month 3 (units)
p3 = cp.Variable(name='p3')  # units produced in month 3 (units)

objective = cp.Minimize(i1 + i2 + 10*p1 + 12*p2 + 11*p3)
constraints = [
    -i1 + p1 == 40,
    i1 - i2 + p2 == 60,
    i2 + p3 == 50,
    p1 <= 55,
    p2 <= 55,
    p3 <= 55,
    p1 >= 0,
    p2 >= 0,
    p3 >= 0,
    i1 >= 0,
    i2 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('p1 =', p1.value)
print('i1 =', i1.value)
print('p2 =', p2.value)
print('i2 =', i2.value)
print('p3 =', p3.value)
```

## The checked answer

The smallest objective value is **1655**, at p1 = 55, i1 = 15, p2 = 45, i2 = 0, p3 = 50. All 11 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
