## The problem

A closed cylindrical can must hold 1000 cm^3. What radius and height use the least metal (surface area)?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 2 \pi \cdot h \cdot r + 2 \pi \cdot r^{2} &&\text{make this as small as possible}\\ \text{s.t.}\;& \pi \cdot h \cdot r^{2} = 1000 &&\text{limit 1: left side must equal 1000}\\ \;& r \ge 0 &&\text{radius of the can (cm) cannot be negative}\\ \;& h \ge 0 &&\text{height of the can (cm) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

# a geometric program: positive variables, solved in log-log form (gp=True)
r = cp.Variable(name='r', pos=True)  # radius of the can (cm)
h = cp.Variable(name='h', pos=True)  # height of the can (cm)

objective = cp.Minimize(6.283185307179586*h*r + 6.283185307179586*r**2)
constraints = [
    3.141592653589793*h*r**2 == 1000,
]
problem = cp.Problem(objective, constraints)
problem.solve(gp=True, solver='CLARABEL')
print(problem.status, problem.value)
print('r =', r.value)
print('h =', h.value)
```

## The checked answer

The smallest objective value is **553.581**, at r = 5.41926, h = 10.8385. All 3 constraints are satisfied (largest violation 3e-13, within the solver's tolerance). **Optimal, numerically checked.**

**Problem class:** Geometric program; **solver:** CLARABEL (geometric program).
