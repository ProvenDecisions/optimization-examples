## The problem

Blend three grains into 1 tonne of feed at least 12% protein and at most 5% fibre. Grain 1: \$200/t, 10% protein, 3% fibre. Grain 2: \$300/t, 18% protein, 4% fibre. Grain 3: \$150/t, 8% protein, 8% fibre. Minimize cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 200 x_{1} + 300 x_{2} + 150 x_{3} &&\text{make this as small as possible}\\ \text{s.t.}\;& x_{1} + x_{2} + x_{3} = 1 &&\text{limit 1: left side must equal 1}\\ \;& 0.1 x_{1} + 0.18 x_{2} + 0.08 x_{3} \ge 0.12 &&\text{limit 2: left side must reach at least 0.12}\\ \;& 0.03 x_{1} + 0.04 x_{2} + 0.08 x_{3} \le 0.05 &&\text{limit 3: left side cannot exceed 0.05}\\ \;& x_{1} \ge 0 &&\text{tonnes of grain 1 in the blend (t) cannot be negative}\\ \;& x_{2} \ge 0 &&\text{tonnes of grain 2 in the blend (t) cannot be negative}\\ \;& x_{3} \ge 0 &&\text{tonnes of grain 3 in the blend (t) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1')  # tonnes of grain 1 in the blend (t)
x2 = cp.Variable(name='x2')  # tonnes of grain 2 in the blend (t)
x3 = cp.Variable(name='x3')  # tonnes of grain 3 in the blend (t)

objective = cp.Minimize(200*x1 + 300*x2 + 150*x3)
constraints = [
    x1 + x2 + x3 == 1,
    0.1*x1 + 0.18*x2 + 0.08*x3 >= 0.120000000000000,
    0.03*x1 + 0.04*x2 + 0.08*x3 <= 0.0500000000000000,
    x1 >= 0,
    x2 >= 0,
    x3 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('x2 =', x2.value)
print('x3 =', x3.value)
```

## The checked answer

The smallest objective value is **216.667**, at x1 = 0.333333, x2 = 0.333333, x3 = 0.333333. All 6 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
