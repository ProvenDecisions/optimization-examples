## The problem

Five jobs of 3, 3, 2, 2 and 2 hours go to two identical machines. Assign them to finish everything as early as possible.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& T &&\text{make as small as possible: time at which all jobs are finished (hours)}\\ \text{s.t.}\;& 3 x_{1} + 3 x_{2} + 2 x_{3} + 2 x_{4} + 2 x_{5} \le T &&\text{limit 1: left side cannot exceed the right side}\\ \;& - 3 x_{1} - 3 x_{2} - 2 x_{3} - 2 x_{4} - 2 x_{5} + 12 \le T &&\text{limit 2: left side cannot exceed the right side}\\ \;& T \ge 0 &&\text{finish time of the last machine, the makespan (hours) cannot be negative}\\ &x_{1}, x_{2}, x_{3}, x_{4}, x_{5} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1', boolean=True)  # 1 if the 3-hour job A is on machine 1, 0 if on machine 2
x2 = cp.Variable(name='x2', boolean=True)  # 1 if the 3-hour job B is on machine 1, 0 if on machine 2
x3 = cp.Variable(name='x3', boolean=True)  # 1 if the first 2-hour job is on machine 1, 0 if on machine 2
x4 = cp.Variable(name='x4', boolean=True)  # 1 if the second 2-hour job is on machine 1, 0 if on machine 2
x5 = cp.Variable(name='x5', boolean=True)  # 1 if the third 2-hour job is on machine 1, 0 if on machine 2
T = cp.Variable(name='T')  # finish time of the last machine, the makespan (hours)

objective = cp.Minimize(T)
constraints = [
    3*x1 + 3*x2 + 2*x3 + 2*x4 + 2*x5 <= T,
    -3*x1 - 3*x2 - 2*x3 - 2*x4 - 2*x5 + 12 <= T,
    T >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('x2 =', x2.value)
print('x3 =', x3.value)
print('x4 =', x4.value)
print('x5 =', x5.value)
print('T =', T.value)
```

## The checked answer

The smallest objective value is **6**, at x1 = 0, x2 = 0, x3 = 1, x4 = 1, x5 = 1, T = 6. All 3 constraints are satisfied (violation ≤0e+00). **Optimal within gap tolerance.**

**Problem class:** MILP; **solver:** HIGHS.
