## The problem

Five items have values 10, 13, 7, 8, 5 and weights 5, 6, 3, 4, 2. The knapsack holds weight 10. Which items maximize total value? Let t1, ..., t5 be 1 if item 1, ..., 5 is taken and 0 if not.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\max\;& 10 t_{1} + 13 t_{2} + 7 t_{3} + 8 t_{4} + 5 t_{5} &&\text{make as large as possible: total value of the items in the knapsack (value units)}\\ \text{s.t.}\;& 5 t_{1} + 6 t_{2} + 3 t_{3} + 4 t_{4} + 2 t_{5} \le 10 &&\text{limit 1: left side cannot exceed 10}\\ &t_{1}, t_{2}, t_{3}, t_{4}, t_{5} \in \{0,1\} &&\text{yes (1) or no (0)}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

t1 = cp.Variable(name='t1', boolean=True)  # 1 if item 1 is taken, 0 otherwise (decision)
t2 = cp.Variable(name='t2', boolean=True)  # 1 if item 2 is taken, 0 otherwise (decision)
t3 = cp.Variable(name='t3', boolean=True)  # 1 if item 3 is taken, 0 otherwise (decision)
t4 = cp.Variable(name='t4', boolean=True)  # 1 if item 4 is taken, 0 otherwise (decision)
t5 = cp.Variable(name='t5', boolean=True)  # 1 if item 5 is taken, 0 otherwise (decision)

objective = cp.Maximize(10*t1 + 13*t2 + 7*t3 + 8*t4 + 5*t5)
constraints = [
    5*t1 + 6*t2 + 3*t3 + 4*t4 + 2*t5 <= 10,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('t1 =', t1.value)
print('t2 =', t2.value)
print('t3 =', t3.value)
print('t4 =', t4.value)
print('t5 =', t5.value)
```

## The checked answer

The largest objective value is **22**, at t1 = 1, t2 = 0, t3 = 1, t4 = 0, t5 = 1. The constraint is satisfied (violation ≤0e+00). **Optimal within gap tolerance.**

**Problem class:** ILP; **solver:** HIGHS.
