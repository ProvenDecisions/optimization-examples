## The problem

Fit y = a + b t to the points (0, 1), (1, 2), (2, 2), (3, 4), (4, 5) minimizing the largest absolute residual.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& d &&\text{make this as small as possible}\\ \text{s.t.}\;& d \ge 1 - a &&\text{limit 1: left side must reach at least the right side}\\ \;& d \ge a - 1 &&\text{limit 2: left side must reach at least the right side}\\ \;& d \ge - a - b + 2 &&\text{limit 3: left side must reach at least the right side}\\ \;& d \ge a + b - 2 &&\text{limit 4: left side must reach at least the right side}\\ \;& d \ge - a - 2 b + 2 &&\text{limit 5: left side must reach at least the right side}\\ \;& d \ge a + 2 b - 2 &&\text{limit 6: left side must reach at least the right side}\\ \;& d \ge - a - 3 b + 4 &&\text{limit 7: left side must reach at least the right side}\\ \;& d \ge a + 3 b - 4 &&\text{limit 8: left side must reach at least the right side}\\ \;& d \ge - a - 4 b + 5 &&\text{limit 9: left side must reach at least the right side}\\ \;& d \ge a + 4 b - 5 &&\text{limit 10: left side must reach at least the right side}\\ \;& d \ge 0 &&\text{largest absolute residual (y units) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

d = cp.Variable(name='d')  # largest absolute residual (y units)
a = cp.Variable(name='a')  # intercept of fitted line (y units)
b = cp.Variable(name='b')  # slope of fitted line (y per unit t), free sign

objective = cp.Minimize(d)
constraints = [
    d >= 1 - a,
    d >= a - 1,
    d >= -a - b + 2,
    d >= a + b - 2,
    d >= -a - 2*b + 2,
    d >= a + 2*b - 2,
    d >= -a - 3*b + 4,
    d >= a + 3*b - 4,
    d >= -a - 4*b + 5,
    d >= a + 4*b - 5,
    d >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('d =', d.value)
print('a =', a.value)
print('b =', b.value)
```

## The checked answer

The smallest objective value is **0.5**, at d = 0.5, a = 0.5, b = 1. All 11 constraints are satisfied (violation ≤0e+00). **Exactly proven optimal.**

**Problem class:** LP; **solver:** HIGHS.
