## The problem

Minimize ||Ax - b||^2 + 0.5 ||x||_1 with A = [[1, 0.5], [0.2, 1], [1, 1], [0.5, -0.3]] and b = (1, 0.8, 1.6, 0.3).

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& \left(0.2 x_{1} + x_{2} - 0.8\right)^{2} + \left(0.5 x_{1} - 0.3 x_{2} - 0.3\right)^{2} + \left(x_{1} + 0.5 x_{2} - 1\right)^{2} + \left(x_{1} + x_{2} - 1.6\right)^{2} + 0.5 \left|{x_{1}}\right| + 0.5 \left|{x_{2}}\right| &&\text{make this as small as possible}\\ \text{s.t.}\;& x_{1} \ge -1000 &&\text{first coefficient of x (unitless, free sign) is at least -1000}\\ \;& x_{2} \ge -1000 &&\text{second coefficient of x (unitless, free sign) is at least -1000}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x1 = cp.Variable(name='x1')  # first coefficient of x (unitless, free sign)
x2 = cp.Variable(name='x2')  # second coefficient of x (unitless, free sign)

objective = cp.Minimize((0.2*x1 + x2 - 0.8)**2 + (0.5*x1 - 0.3*x2 - 0.3)**2 + (x1 + 0.5*x2 - 1)**2 + (x1 + x2 - 1.6)**2 + 0.5*cp.abs(x1) + 0.5*cp.abs(x2))
constraints = [
    x1 >= -1000,
    x2 >= -1000,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('x1 =', x1.value)
print('x2 =', x2.value)
```

## The checked answer

The smallest objective value is **0.753283**, at x1 = 0.763303, x2 = 0.58841. All 2 constraints are satisfied (violation ≤0e+00). **Optimal per solver.**
