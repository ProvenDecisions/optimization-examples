## The problem

Liabilities of \$100k, \$120k and \$150k are due at the ends of years 1, 2 and 3. Bond A (price \$95) pays \$100 at year 1. Bond B (price \$90) pays \$5 at years 1 and 2 and \$105 at year 3. Bond C (price \$92) pays \$6 at year 1 and \$106 at year 2. Surplus cash is carried forward at 0% interest. Buy fractional units (in thousands) to meet every liability at minimum cost.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 95 x_{A} + 90 x_{B} + 92 x_{C} &&\text{make this as small as possible}\\ \text{s.t.}\;& - s_{1} + 100 x_{A} + 5 x_{B} + 6 x_{C} = 100 &&\text{limit 1: left side must equal 100}\\ \;& s_{1} - s_{2} + 5 x_{B} + 106 x_{C} = 120 &&\text{limit 2: left side must equal 120}\\ \;& s_{2} + 105 x_{B} = 150 &&\text{limit 3: left side must equal 150}\\ \;& x_{A} \ge 0 &&\text{units of Bond A bought (thousands of bonds) cannot be negative}\\ \;& x_{B} \ge 0 &&\text{units of Bond B bought (thousands of bonds) cannot be negative}\\ \;& x_{C} \ge 0 &&\text{units of Bond C bought (thousands of bonds) cannot be negative}\\ \;& s_{1} \ge 0 &&\text{surplus cash carried from year 1 to year 2 (\$k) cannot be negative}\\ \;& s_{2} \ge 0 &&\text{surplus cash carried from year 2 to year 3 (\$k) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

xA = cp.Variable(name='xA')  # units of Bond A bought (thousands of bonds)
xB = cp.Variable(name='xB')  # units of Bond B bought (thousands of bonds)
xC = cp.Variable(name='xC')  # units of Bond C bought (thousands of bonds)
s1 = cp.Variable(name='s1')  # surplus cash carried from year 1 to year 2 ($k)
s2 = cp.Variable(name='s2')  # surplus cash carried from year 2 to year 3 ($k)

objective = cp.Minimize(95*xA + 90*xB + 92*xC)
constraints = [
    -s1 + 100*xA + 5*xB + 6*xC == 100,
    s1 - s2 + 5*xB + 106*xC == 120,
    s2 + 105*xB == 150,
    xA >= 0,
    xB >= 0,
    xC >= 0,
    s1 >= 0,
    s2 >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('xA =', xA.value)
print('xB =', xB.value)
print('xC =', xC.value)
print('s1 =', s1.value)
print('s2 =', s2.value)
```

**Problem class:** LP; **solver:** HIGHS.
