## The problem

Find the smallest circle that contains the points (0, 0), (4, 0) and (0, 3).

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& r &&\text{make this as small as possible}\\ \text{s.t.}\;& c_{x}^{2} + c_{y}^{2} \le r^{2} &&\text{limit 1: left side cannot exceed the right side}\\ \;& c_{y}^{2} + \left(c_{x} - 4\right)^{2} \le r^{2} &&\text{limit 2: left side cannot exceed the right side}\\ \;& c_{x}^{2} + \left(c_{y} - 3\right)^{2} \le r^{2} &&\text{limit 3: left side cannot exceed the right side}\\ \;& r \ge 0 &&\text{radius of the circle (length) cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

cx = cp.Variable(name='cx')  # x-coordinate of circle center (length)
cy = cp.Variable(name='cy')  # y-coordinate of circle center (length)
r = cp.Variable(name='r')  # radius of the circle (length)

objective = cp.Minimize(r)
constraints = [
    cp.norm(cp.hstack([cx, cy]), 2) <= r,
    cp.norm(cp.hstack([cy, cx - 4]), 2) <= r,
    cp.norm(cp.hstack([cx, cy - 3]), 2) <= r,
    r >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('cx =', cx.value)
print('cy =', cy.value)
print('r =', r.value)
```

## The checked answer

The smallest objective value is **2.5**, at cx = 1.99986, cy = 1.49981, r = 2.5. All 4 constraints are satisfied (largest violation 2e-08, within the solver's tolerance). **Optimal, numerically checked.**

**Problem class:** Conic (convex); **solver:** CLARABEL.
