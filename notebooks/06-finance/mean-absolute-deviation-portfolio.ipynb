## The problem

Two funds returned (%) A: 4, -2, 3, 1 and B: 1, 2, -1, 2 in four equally likely scenarios. Fully invested, no shorts. Minimize the mean absolute deviation of the portfolio return from its mean.

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 0.625 \left|{w_{a}}\right| + 0.25 \left|{0.5 w_{a} - w_{b}}\right| + 0.25 \left|{1.5 w_{a} - 2 w_{b}}\right| + 0.25 \left|{3.5 w_{a} - w_{b}}\right| &&\text{make this as small as possible}\\ \text{s.t.}\;& w_{a} + w_{b} = 1 &&\text{limit 1: left side must equal 1}\\ \;& w_{a} \ge 0 &&\text{fraction of the portfolio invested in fund A cannot be negative}\\ \;& w_{b} \ge 0 &&\text{fraction of the portfolio invested in fund B cannot be negative}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

wa = cp.Variable(name='wa')  # fraction of the portfolio invested in fund A
wb = cp.Variable(name='wb')  # fraction of the portfolio invested in fund B

objective = cp.Minimize(0.625*cp.abs(wa) + 0.25*cp.abs(0.5*wa - wb) + 0.25*cp.abs(1.5*wa - 2*wb) + 0.25*cp.abs(3.5*wa - wb))
constraints = [
    wa + wb == 1,
    wa >= 0,
    wb >= 0,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='HIGHS')
print(problem.status, problem.value)
print('wa =', wa.value)
print('wb =', wb.value)
```

## The checked answer

The smallest objective value is **0.611111**, at wa = 0.222222, wb = 0.777778. All 3 constraints are satisfied (violation ≤0e+00). **Optimal, numerically checked.**
