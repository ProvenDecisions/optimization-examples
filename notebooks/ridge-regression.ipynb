## The problem

Minimize ||Ax - b||^2 + 0.5 ||x||^2 with A = [[1, 0], [1, 1], [1, 2], [1, 3]] and b = (1, 2, 2, 4).

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& 0.5 x_{0}^{2} + 0.5 x_{1}^{2} + \left(x_{0} - 1\right)^{2} + \left(x_{0} + x_{1} - 2\right)^{2} + \left(x_{0} + 2 x_{1} - 2\right)^{2} + \left(x_{0} + 3 x_{1} - 4\right)^{2} &&\text{make this as small as possible}\\ \text{s.t.}\;& x_{0} \ge -1000 &&\text{intercept coefficient (unitless) is at least -1000}\\ \;& x_{1} \ge -1000 &&\text{slope coefficient (unitless) is at least -1000}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

x0 = cp.Variable(name='x0')  # intercept coefficient (unitless)
x1 = cp.Variable(name='x1')  # slope coefficient (unitless)

objective = cp.Minimize(0.5*x0**2 + 0.5*x1**2 + (x0 - 1)**2 + (x0 + x1 - 2)**2 + (x0 + 2*x1 - 2)**2 + (x0 + 3*x1 - 4)**2)
constraints = [
    x0 >= -1000,
    x1 >= -1000,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('x0 =', x0.value)
print('x1 =', x1.value)
```

## The checked answer

The smallest objective value is **1.46154**, at x0 = 0.769231, x1 = 0.923077. All 2 constraints are satisfied (violation ≤0e+00). **Optimal, numerically checked.**

**Problem class:** Convex QP; **solver:** CLARABEL.
