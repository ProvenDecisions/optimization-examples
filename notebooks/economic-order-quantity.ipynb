## The problem

A shop sells 1200 units a year. Each order costs \$50 to place and holding one unit for a year costs \$3. What order size minimizes yearly ordering plus holding cost?

## The model

Each line has a plain-language note.

$$
\begin{aligned}\min\;& \frac{3 q}{2} + \frac{60000}{q} &&\text{make this as small as possible}\\ \text{s.t.}\;& q \ge 0.001 &&\text{order size (units per order) is at least 0.001}\end{aligned}
$$

## Solve it with cvxpy

Run the cell (in Colab, `cvxpy` is installed by the first line).

```python
%pip install -q cvxpy highspy
```

```python
import cvxpy as cp

q = cp.Variable(name='q')  # order size (units per order)

objective = cp.Minimize(60000*cp.inv_pos(q) + 3*q/2)
constraints = [
    q >= 0.001,
]
problem = cp.Problem(objective, constraints)
problem.solve(solver='CLARABEL')
print(problem.status, problem.value)
print('q =', q.value)
```

## The checked answer

You asked for: the optimal order size q. The smallest objective value is **600**, at q = 200.001. The constraint is satisfied exactly. **Optimal, numerically checked.**

**Problem class:** Conic (convex); **solver:** CLARABEL.
